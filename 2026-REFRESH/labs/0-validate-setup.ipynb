# Environment Setup Validation

This notebook validates that all required environment variables are properly configured for the AI Red Teaming lab exercises using the **NEW Azure AI Foundry SDK (2026)**.

## What This Validates

✅ Azure subscription and resource configuration  
✅ Azure AI Foundry project settings  
✅ Model deployment and agent configuration  
✅ Lab-specific requirements (OpenAI API, optional variables)

## Prerequisites

- Run `./2026-REFRESH/infra/setup-env.sh` to configure your `.env` file
- Ensure `.env` file exists in the **repository root** directory
- Azure CLI authentication is active (`az login`)
- Using Python 3.10, 3.11, or 3.12 (PyRIT requirement)


## Import Required Libraries

```python
import os
from pathlib import Path
from dotenv import load_dotenv

# Load environment variables from .env file in REPOSITORY ROOT
# Navigate up from labs/ to 2026-REFRESH/ to repo root
env_path = Path(__file__).parent.parent.parent / ".env" if "__file__" in globals() else Path("../../.env")
load_dotenv(dotenv_path=env_path)

print(f"🔍 Loading environment from: {env_path.absolute()}")
print(f"   Exists: {env_path.exists()}")
if not env_path.exists():
    print(f"\n⚠️  WARNING: .env file not found!")
    print(f"   Run: ./2026-REFRESH/infra/setup-env.sh")
print()

# Tracking validation results
missing_vars = []
valid_vars = []
warnings = []
```

## 1. Validate Azure Subscription & Resource Group

Check core Azure subscription and resource configuration.

```python
azure_core_vars = {
    "AZURE_SUBSCRIPTION_ID": "Your Azure subscription ID",
    "AZURE_RESOURCE_GROUP": "Resource group containing your Foundry project",
    "AZURE_LOCATION": "Azure region (eastus, westus, westeurope, swedencentral recommended)"
}

print("🔍 Checking Azure Core Configuration...")
print("=" * 70)
for var, description in azure_core_vars.items():
    value = os.getenv(var)
    if value:
        # Mask sensitive values - only show if SET
        if var == "AZURE_SUBSCRIPTION_ID":
            display_value = f"{value[:4]}***"
        else:
            display_value = value
        print(f"✅ {var}")
        print(f"   {description}")
        print(f"   Value: {display_value}\n")
        valid_vars.append(var)
    else:
        print(f"❌ {var}")
        print(f"   {description}")
        print(f"   Status: NOT SET\n")
        missing_vars.append(var)
```

## 2. Validate Azure AI Project Configuration

Check AI Foundry project and service configuration.

```python
ai_project_vars = {
    "PROJECT_ENDPOINT": "Foundry project endpoint (primary format)",
    "AZURE_AI_PROJECT": "Project name",
    "AZURE_AI_SERVICE": "AI Service connection name (optional)"
}

# Check for alternative endpoint format
alt_endpoint = os.getenv("AZURE_AI_PROJECT_ENDPOINT")

print("🔍 Checking Azure AI Foundry Project Configuration...")
print("=" * 70)

for var, description in ai_project_vars.items():
    value = os.getenv(var)
    if value:
        # Mask endpoints for security
        if "ENDPOINT" in var:
            display_value = "***SET***"
        else:
            display_value = value
        print(f"✅ {var}")
        print(f"   {description}")
        print(f"   Value: {display_value}\n")
        valid_vars.append(var)
    elif var == "AZURE_AI_SERVICE":
        # Optional variable
        print(f"⚠️  {var} (optional)")
        print(f"   {description}")
        print(f"   Status: NOT SET (OK if not using AI Services)\n")
    else:
        print(f"❌ {var}")
        print(f"   {description}")
        print(f"   Status: NOT SET\n")
        missing_vars.append(var)

# Check for alternative endpoint format
if alt_endpoint and "PROJECT_ENDPOINT" not in valid_vars:
    print(f"✅ AZURE_AI_PROJECT_ENDPOINT (alternative format)")
    print(f"   Found alternative endpoint format: ***SET***\n")
    valid_vars.append("PROJECT_ENDPOINT")  # Mark as satisfied
```

## 3. Validate Model & Agent Configuration

Check model deployment and agent settings for red teaming.

```python
model_agent_vars = {
    "AZURE_AI_DEPLOYMENT_NAME": "Model deployment name (e.g., gpt-4)",
    "AZURE_AI_AGENT_NAME": "Agent name for red teaming (e.g., my-first-agent)"
}

# Check for alternative variable names
alt_deployment = os.getenv("AZURE_AI_MODEL_DEPLOYMENT_NAME")

print("🔍 Checking Model & Agent Configuration...")
print("=" * 70)

for var, description in model_agent_vars.items():
    value = os.getenv(var)
    if value:
        print(f"✅ {var}")
        print(f"   {description}")
        print(f"   Value: {value}\n")
        valid_vars.append(var)
    else:
        print(f"❌ {var}")
        print(f"   {description}")
        print(f"   Status: NOT SET\n")
        missing_vars.append(var)

# Check for alternative deployment name
if alt_deployment and "AZURE_AI_DEPLOYMENT_NAME" not in valid_vars:
    print(f"✅ AZURE_AI_MODEL_DEPLOYMENT_NAME (alternative format)")
    print(f"   Found alternative deployment name: {alt_deployment}\n")
    valid_vars.append("AZURE_AI_DEPLOYMENT_NAME")  # Mark as satisfied
```

## 4. Validate Lab 2 - Azure OpenAI Configuration

Check Azure OpenAI direct API access configuration (used in Lab 2 - Model Scanning).

```python
# Lab 2 requires direct Azure OpenAI API access with alternative variable names
openai_vars = {
    "REQUIRED": {
        "MODEL_ENDPOINT": "Azure OpenAI endpoint (https://<name>.openai.azure.com)",
        "MODEL_API_KEY": "Azure OpenAI API key",
        "MODEL_DEPLOYMENT_NAME": "Model deployment name",
        "AZURE_OPENAI_API_VERSION": "Azure OpenAI API version (e.g., 2024-10-01-preview)"
    },
    "ALTERNATIVE": {
        "AZURE_OPENAI_ENDPOINT": "Alternative endpoint variable",
        "AZURE_OPENAI_API_KEY": "Alternative API key variable",
        "AZURE_OPENAI_DEPLOYMENT": "Alternative deployment variable"
    }
}

print("🔍 Checking Lab 2 - Azure OpenAI Direct API Configuration...")
print("=" * 70)
print("Note: Lab 2 requires EITHER the MODEL_* OR AZURE_OPENAI_* variables\n")

lab2_ready = True

# Check required variables and their alternatives
for var, description in openai_vars["REQUIRED"].items():
    value = os.getenv(var)
    
    # For API version, check if it exists
    if var == "AZURE_OPENAI_API_VERSION":
        if value:
            print(f"✅ {var}")
            print(f"   {description}")
            print(f"   Value: {value}\n")
            valid_vars.append(var)
        else:
            print(f"⚠️  {var}")
            print(f"   {description}")
            print(f"   Status: NOT SET (will use default)\n")
            warnings.append(f"{var} - will use SDK default")
        continue
    
    # Check for alternatives for other required vars
    alt_var = None
    if var == "MODEL_ENDPOINT":
        alt_var = "AZURE_OPENAI_ENDPOINT"
    elif var == "MODEL_API_KEY":
        alt_var = "AZURE_OPENAI_API_KEY"
    elif var == "MODEL_DEPLOYMENT_NAME":
        alt_var = "AZURE_OPENAI_DEPLOYMENT"
    
    alt_value = os.getenv(alt_var) if alt_var else None
    
    if value:
        # Mask sensitive values - keys and endpoints
        if var.endswith("_KEY"):
            display = "***MASKED***"
        elif "ENDPOINT" in var:
            display = "***SET***"
        else:
            display = value
        print(f"✅ {var}")
        print(f"   {description}")
        print(f"   Value: {display}\n")
        valid_vars.append(var)
    elif alt_value:
        # Mask sensitive values for alternatives too
        if alt_var.endswith("_KEY"):
            display = "***MASKED***"
        elif "ENDPOINT" in alt_var:
            display = "***SET***"
        else:
            display = alt_value
        print(f"✅ {alt_var} (alternative for {var})")
        print(f"   {description}")
        print(f"   Value: {display}\n")
        valid_vars.append(var)
    else:
        print(f"⚠️  {var} (and alternative {alt_var})")
        print(f"   {description}")
        print(f"   Status: NOT SET\n")
        warnings.append(f"{var} or {alt_var} needed for Lab 2")
        lab2_ready = False

if lab2_ready:
    print("✅ Lab 2 configuration is complete!")
else:
    print("⚠️  Lab 2 configuration is incomplete - some variables missing")
```

## 5. Validation Summary

Review overall configuration status and next steps.

```python
print("\n" + "="*70)
print("📊 VALIDATION SUMMARY")
print("="*70 + "\n")

print(f"✅ Valid variables: {len(valid_vars)}")
print(f"❌ Missing required variables: {len(missing_vars)}")
print(f"⚠️  Warnings: {len(warnings)}\n")

# Determine lab readiness
core_ready = all(var not in missing_vars for var in ["AZURE_SUBSCRIPTION_ID", "AZURE_RESOURCE_GROUP", "PROJECT_ENDPOINT", "AZURE_AI_PROJECT"])
lab1_ready = core_ready and "AZURE_AI_AGENT_NAME" not in missing_vars
lab2_ready = "MODEL_ENDPOINT" in valid_vars or "AZURE_OPENAI_ENDPOINT" in os.environ
lab3_ready = core_ready

print("📋 Lab Readiness Status:")
print("="*70)
print(f"{'Lab 1 - Agent Red Teaming:':<35} {'✅ READY' if lab1_ready else '❌ NOT READY'}")
print(f"{'Lab 2 - Model Red Teaming:':<35} {'✅ READY' if lab2_ready else '⚠️  INCOMPLETE'}")
print(f"{'Lab 3 - Cloud Red Teaming:':<35} {'✅ READY' if lab3_ready else '❌ NOT READY'}")
print(f"{'Lab 4 - Advanced Red Teaming:':<35} {'✅ READY' if lab3_ready else '❌ NOT READY'}")
print()

if missing_vars:
    print("❌ Missing REQUIRED environment variables:")
    print("="*70)
    for var in missing_vars:
        print(f"   • {var}")
    print()

if warnings:
    print("⚠️  Warnings (may affect specific labs):")
    print("="*70)
    for warn in warnings:
        print(f"   • {warn}")
    print()

if missing_vars:
    print("💡 Next Steps:")
    print("="*70)
    print("   1. Copy .env.sample to .env:")
    print("      cp .env.sample .env")
    print()
    print("   2. Fill in your values in .env file")
    print()
    print("   3. Or run the setup script:")
    print("      ./infra/2-setup-env.sh")
    print()
    print("   4. Make sure you're logged in to Azure:")
    print("      az login")
    print()
    print("   5. Reload this notebook to re-validate")
    print()
elif warnings and not lab2_ready:
    print("✅ Core environment is properly configured!")
    print()
    print("⚠️  Lab 2 (Model Red Teaming) requires additional Azure OpenAI variables.")
    print("   Set either MODEL_* or AZURE_OPENAI_* variables in your .env file.")
    print()
    print("📚 You can proceed with:")
    print("   • Lab 1: Agent Red Teaming")
    print("   • Lab 3: Cloud Red Teaming")
    print("   • Lab 4: Advanced Red Teaming")
    print()
else:
    print("🎉 All environment variables are properly configured!")
    print()
    print("✅ Your environment is ready for all lab exercises:")
    print()
    print("📚 Available Labs:")
    print("="*70)
    print("   1. Lab 1 - Agent Red Teaming")
    print("      Scan AI agents for vulnerabilities and safety issues")
    print("      📁 labs/1-scan-agent/")
    print()
    print("   2. Lab 2 - Model Red Teaming")
    print("      Test model deployments with custom prompts and strategies")
    print("      📁 labs/2-scan-model/")
    print()
    print("   3. Lab 3 - Cloud Red Teaming")
    print("      Use cloud-based AI Red Teaming Agent for scalable testing")
    print("      📁 labs/3-scan-cloud/")
    print()
    print("   4. Lab 4 - Advanced Red Teaming")
    print("      Complete end-to-end red teaming of Zava's Cora assistant")
    print("      📁 labs/4-scan-more/")
    print()

print("="*70)
print("🔐 Security Reminder:")
print("   • Never commit .env files to version control")
print("   • Rotate API keys regularly")
print("   • Use managed identities in production")
print("="*70)
```
